## Завдання
Згенерувати 100 значень $x$ у межах від 0 до 20, обчислити $y = 8x + 2 + N(0, 2)$, побудувати діаграму розсіювання, навчити модель лінійної регресії та додати її лінію на графік.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

VARIANT = 8
SAMPLE_COUNT = 100
RANDOM_SEED = 42
```

```python
generator = np.random.default_rng(RANDOM_SEED)
x = generator.uniform(0, 20, SAMPLE_COUNT)
y = VARIANT * x + 2 + generator.normal(0, 2, SAMPLE_COUNT)
x.shape, y.shape, (x.min(), x.max())
```

```python
model = LinearRegression()
model.fit(x.reshape(-1, 1), y)
predictions = model.predict(x.reshape(-1, 1))
print(f"Коефіцієнт: {model.coef_[0]:.4f}")
print(f"Вільний член: {model.intercept_:.4f}")
print(f"R²: {r2_score(y, predictions):.4f}")
```

```python
x_line = np.linspace(0, 20, 200)
y_line = model.predict(x_line.reshape(-1, 1))

plt.figure(figsize=(9, 6))
plt.scatter(x, y, alpha=0.72, color="#2563eb", label="Згенеровані дані")
plt.plot(x_line, y_line, color="#dc2626", linewidth=2.5, label="Лінійна регресія")
plt.title("Лінійна регресія: варіант 8")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.28)
plt.legend()
plt.show()
```

## Висновок
Модель відновила залежність, близьку до заданої формули $y = 8x + 2$, а високе значення $R^2$ підтверджує якість апроксимації.
